Distributed indexing involves dividing and distributing the indexing process across multiple nodes or machines. This approach offers several advantages:

- Scalability: Handles large-scale data by distributing the workload.
- Performance: Improves performance by parallelizing the indexing process.
- Fault Tolerance: Ensures system resilience by continuing operations on remaining nodes if one fails.
- Load Balancing: Balances the load across different nodes to prevent bottlenecks.

To apply distributed indexing:
- Data Partitioning: Divide the data into smaller chunks that can be processed independently.
- Indexing Nodes: Set up multiple nodes to handle different parts of the data.
- Parallel Processing: Use frameworks like Apache Spark or Hadoop to index data chunks simultaneously.
- Aggregation: Aggregate results from all nodes to create a unified index.
- Monitoring and Maintenance: Continuously monitor and maintain the system for optimal performance and reliability.

By leveraging distributed indexing, you can enhance the scalability, performance, and reliability of your information retrieval system, making it more efficient and robust.

```python
from util.load_lotr import lotr_chapters
from collections import defaultdict

def map_function(chapter):
    # Tokenize the chapter into words
    words = chapter.lower().split()
    # Emit each word with a count of 1
    for word in words:
        yield (word, 1)

def reduce_function(word, counts):
    # Sum the counts for each word
    yield (word, sum(counts))

def main():
    chapters = lotr_chapters()  # Load the chapters

    # Map Phase: Process each chapter in parallel
    mapped_data = []
    for chapter in chapters:
        mapped_data.extend(map_function(chapter))

    # Shuffle Phase: Group the mapped data by word
    shuffled_data = defaultdict(list)
    for word, count in mapped_data:
        shuffled_data[word].append(count)

    # Reduce Phase: Aggregate the counts for each word
    reduced_data = []
    for word, counts in shuffled_data.items():
        reduced_data.extend(reduce_function(word, counts))

    # Output the results
    for word, total in reduced_data:
        print(f"{word}: {total}")

if __name__ == "__main__":
    main()
```

```python
def assign_tasks(tasks, compute_units):
    # Sort tasks in descending order to handle larger tasks first
    tasks.sort(reverse=True)
    # Sort compute units in descending order to handle larger capacities first
    compute_units.sort(reverse=True)

    assignments = {i: [] for i in range(len(compute_units))}

    for task in tasks:
        for i, capacity in enumerate(compute_units):
            if capacity >= task:
                assignments[i].append(task)
                compute_units[i] -= task
                break

    return assignments

# Example usage
tasks = [5, 3, 7, 2, 4, 6]
compute_units = [8, 6, 4, 5]

assignments = assign_tasks(tasks, compute_units)
for unit, assigned_tasks in assignments.items():
    print(f"Compute Unit {unit}: {assigned_tasks}")
```

**Explanation:**
- Sorting: The tasks and compute units are sorted in descending order to handle larger tasks and capacities first. This helps in efficiently utilizing the available capacities.
- Assignment: For each task, the algorithm checks the compute units in order to find the first one that can accommodate the task without exceeding its capacity. The task is then assigned to that compute unit, and the capacity of the compute unit is reduced by the workload of the task.
- Output: The assignments are printed, showing which tasks are assigned to each compute unit.
