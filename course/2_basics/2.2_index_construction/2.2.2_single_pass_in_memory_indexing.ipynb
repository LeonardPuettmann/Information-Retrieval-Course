SPIMI, or Single Pass In-Memory Indexing, is a simple and efficient method for creating an inverted index from a large document collection. It works by processing the documents in a single pass, building an in-memory dictionary of terms and their postings lists, and writing the dictionary to disk when it becomes too large to fit in memory. This approach is efficient because it minimizes the number of passes over the data and only requires enough memory to hold the current dictionary. SPIMI is widely used in information retrieval systems, such as search engines, to create an inverted index from a large document collection.

The code implements a simple version of the Single Pass In-Memory Indexing (SPIMI) algorithm for creating an inverted index. The algorithm processes a stream of tokens (in this case, words from the Lord of the Rings chapters) and builds an in-memory dictionary of terms and their postings lists. The postings lists are simplified for demonstration purposes, with each posting being a doubled version of the previous one. The final sorted dictionary is then written to an output file. This approach is efficient for creating an inverted index from a large document collection, as it processes the data in a single pass and only requires enough memory to hold the current dictionary.

```python
from util.load_lotr import lotr_chapters
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')

def spimi_invert(doc_ids):
    output_file = open('blocks/output.txt', 'w')
    dictionary = {}

    # Get the set of English stop words
    stop_words = set(stopwords.words('english'))

    for doc_id, chapter in doc_ids.items():
        for token in chapter.split():
            term = token.lower()  # Convert term to lowercase for case-insensitive comparison

            if term not in stop_words:  # Check if the term is not a stop word
                if term not in dictionary:
                    dictionary[term] = []
                    postings_list = dictionary[term]
                else:
                    postings_list = dictionary[term]

                if len(postings_list) == 0 or postings_list[-1] != doc_id:  # Check if the last posting is not the current doc_id
                    postings_list.append(doc_id)  # Append the current doc_id

    sorted_terms = sorted(dictionary.keys())
    for term in sorted_terms:
        postings_list = dictionary[term]
        output_file.write(f"{term}: {postings_list}\n")

    output_file.close()
    return dictionary

# Load the chapters from the Lord of the Rings books
chapters = lotr_chapters()  # This loads a list of strings
chapters = [c.lower().strip().replace(",", "").replace(".", "").replace("!", "").replace("?", "").replace("-", "").replace(";", "").replace(":", "") for c in chapters]

# Create a unique doc_id for all items in the chapters list
doc_ids = {i: chapter for i, chapter in enumerate(chapters, start=1)}

# Call the spimi_invert function with the doc_ids
output_file = spimi_invert(doc_ids)
```

### Solution

```python
def spimi_invert(documents):
    dictionary = {}

    for doc_id, document in enumerate(documents):
        for word in document.split():
            term = word.lower()  # Normalize the term to lowercase
            if term not in dictionary:
                dictionary[term] = []
            dictionary[term].append(doc_id)

    # Sort the dictionary by term
    sorted_terms = sorted(dictionary.keys())

    # Write the index to a file
    with open('index.txt', 'w') as output_file:
        for term in sorted_terms:
            postings_list = dictionary[term]
            output_file.write(f"{term}: {postings_list}\n")

# Example usage
documents = [
    "The quick brown fox jumps over the lazy dog",
    "The lazy dog sleeps in the sun",
    "The quick brown fox is quick"
]

spimi_invert(documents)
```
